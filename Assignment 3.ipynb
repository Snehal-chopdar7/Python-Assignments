```python
# Python Assignment 3
def is_right_triangle():
  side1=int(input("Enter side1:"))
  side2=int(input("Enter side2:"))
  side3=int(input("Enter side3:"))
  if side1**2 + side2**2 == side3**2:
    print("it is a right triangle")
  else:
    print("it is not right triangle")
is_right_triangle()
```
