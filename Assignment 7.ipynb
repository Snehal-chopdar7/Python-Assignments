```python
import pandas as pd
data = {'Student':['Amit','Bhavana','Chetan','Diya','Esha','Farhan'],
        'Maths':[75,82,68,90,72,85],
        'Python':[80,78,70,95,80,85],
        'Attendence':[85,90,78,95,80,92],
        'City':['pune','Mumbai','Pune','Nashik','Mumbai','Pune']}

df_setup = pd.DataFrame(data)
df_setup.to_csv("sample_data_assignment_07.csv", index=False)
print("Data file created successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV dataset
df = pd.read_csv("sample_data_assignment_07.csv")

# Display the dataset
print("Dataset:")
print(df)

# Line Plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Correlation Matrix and Heatmap
correlation = df[["Maths", "Python", "Attendence"]].corr()
print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd
data = {'Student':['Sana','Bob','Charlie','David','Eva'],
        'City':['New York','Los Angeles','New York','Chicago','Los Angeles'],
        'Maths':[85,78,92,65,88],
        'Python':[90,85,95,70,78],
        'Attendance':[92,88,95,80,91]}

df_setup = pd.DataFrame(data)
df_setup.to_csv("sample_data_assignment_07.csv", index=False)
print("Data file created successfully!")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Get the CSV file name from the user
df = pd.read_csv("sample_data_assignment_07.csv")

# Display the dataset
print("Dataset:")
print(df)

# 3. Calculate and display averages
print("Average Maths:", df["Maths"].mean())
print("Average Python:", df["Python"].mean())
print("Average Attendance:", df["Attendance"].mean())

# --- Plot 1: Histogram of Python Marks ---
plt.figure()  # Opens a new window for the plot
plt.hist(df["Python"])
plt.title("Python Marks Histogram")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# --- Plot 2: Box Plot for Maths and Python ---
plt.figure()
plt.boxplot([df["Maths"], df["Python"]], labels=["Maths", "Python"])
plt.title("Maths vs Python Box Plot")
plt.ylabel("Marks")
plt.show()

# --- Plot 3: Pie Chart for Cities ---
plt.figure()
city_counts = df["City"].value_counts()  # Counts how many students in each city
plt.pie(city_counts, labels=city_counts.index, autopct="%1.1f%%")
plt.title("Students by City")
plt.legend(title="Cities")
plt.show()
```
